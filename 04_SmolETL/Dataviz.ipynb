### 4. Smol ETL
Objectifs : construire un POC d'ETL avec DuckDB
Use case métier : disposer d'une solution rapide d'analytics en local

```python
import duckdb as d
import pandas as pd
```

```python
#Connect to DB
con = d.connect('NYC_yellow_taxi.duckdb')
```

```python
#Plot number of trips per day
trips_per_day = con.execute("""
    SELECT trip_date, SUM(total_trips) as daily_trips
    FROM facts_trips_daily
    GROUP BY trip_date
    ORDER BY trip_date
""").df()

trips_per_day.plot(x='trip_date', y='daily_trips', kind='line', title='Number of Trips per Day')
```

```python
#Connect to DB
con.close()
```
